```python
# Load data from control run 
#lfric_path_control = '/g/data/access/projects/access/lfric/workshop-2024/canned-output-gal/expt-0/cycle/19881001T0000Z/lfric_stream_m.nc'
lfric_path_control = '/PATH/TO/LFRIC/CONTROL/DATA/'
# Load data from experimental run
#lfric_path = '/g/data/access/projects/access/lfric/workshop-2024/canned-output-gal/expt-1/cycle/19881001T0000Z/lfric_stream_m.nc'
lfric_path = '/PATH/TO/LFRIC/DATA/'
```

```python
# load the lfric data as a cube and extract varaible
from iris.experimental.ugrid.load import PARSE_UGRID_ON_LOAD
with iris.experimental.ugrid.PARSE_UGRID_ON_LOAD.context():
    lfric_rho_control = iris.load_cube(lfric_path_control, 't1p5m_6hourly') # chose the variable to plot here
print(lfric_rho_control)
```

```python
# load the lfric data as a cube and extract varaible
with iris.experimental.ugrid.PARSE_UGRID_ON_LOAD.context():
    lfric_rho = iris.load_cube(lfric_path, 't1p5m_6hourly')
print(lfric_rho)
```

```python
# make difference cube
lfric_diff = lfric_rho - lfric_rho_control
```

```python
#select first timestep from the data
lfric_rho_t0_control = lfric_rho_control[29]
lfric_rho_t0 = lfric_rho[29]
lfric_rho_t0_diff = lfric_diff[29]
```

```python
#Transorm LFRic cuve to pyvista object
lfric_pv_control = pv_from_lfric_cube(lfric_rho_t0_control)
lfric_pv = pv_from_lfric_cube(lfric_rho_t0)
lfric_pv_diff = pv_from_lfric_cube(lfric_rho_t0_diff)
```

```python
# Plot the data using pyvista
plotter = GeoPlotter(shape=(1, 3))
plotter.subplot(0, 0)
plotter.add_coastlines()
plotter.add_title('control')
plotter.add_base_layer(texture=gv.natural_earth_1())
plotter.add_mesh(lfric_pv_control)
```

```python
plotter.subplot(0, 1)
plotter.add_coastlines()
plotter.add_title('experiment')
plotter.add_base_layer(texture=gv.natural_earth_1())
plotter.add_mesh(lfric_pv)
```

```python
plotter.subplot(0, 2)
plotter.add_coastlines()
plotter.add_title('diff')
plotter.add_base_layer(texture=gv.natural_earth_1())
plotter.add_mesh(lfric_pv_diff)
```

```python
# Make left+right move together
plotter.link_views()

# Try to rationalise the global view a little.
plotter.view_xz()

plotter.show()
```
